# Embeddingi

Embedding to **wektor liczb o stałej długości**, który reprezentuje znaczenie tekstu. Teksty o podobnym znaczeniu dają wektory leżące blisko siebie – dzięki temu możemy wyszukiwać fragmenty semantycznie podobne do pytania użytkownika, co jest podstawą systemów RAG.

## Różnice między modelami
Wybierając model do embeddingów, bierzemy pod uwagę:
- **jakość** – szczególnie dla języka, w którym są nasze dokumenty
- **szybkość i koszt** – model lokalny (darmowy, ale wymaga zasobów) vs. API (płatne za tokeny)
- **rozmiar wektora** – większy zwykle oznacza lepszą jakość, ale też więcej obliczeń i miejsca w bazie
- **maksymalną długość tekstu** – dłuższy tekst zostanie ucięty (więcej o tym niżej)

## HuggingFace

HuggingFace to platforma z modelami open source, które możemy pobrać i uruchomić lokalnie. W LangChain korzystamy z nich przez klasę `HuggingFaceEmbeddings`.

Nazwa repozytorium ma postać `<organizacja>/<model>`, np. `sentence-transformers/all-mpnet-base-v2`. Z nazwy modelu często da się odczytać:
- **język** – brak oznaczenia zwykle oznacza angielski, `multilingual` – wiele języków
- **architekturę** – np. `MiniLM`, `mpnet`, `roberta`
- **rozmiar** – np. `L6` (6 warstw), `base`, `large`

https://huggingface.co/sentence-transformers/all-mpnet-base-v2

Popularny model dla języka angielskiego. Zwraca wektor o długości 768.

### Maksymalna długość tekstu

Każdy model ma limit tokenów, które przetwarza (`max_seq_length`). **Tekst dłuższy niż limit jest po cichu ucinany** – reszta w ogóle nie wpływa na embedding. Dlatego rozmiar chunków trzeba dobrać do modelu.

```python
embeddings
```

```python
embeddings._client
```

```python
print("Rozmiar wektora:", embeddings._client.get_embedding_dimension())
print("Maks. liczba tokenów w dokumencie:", embeddings._client.max_seq_length)
```

```python
text
```

```python
long_text = text * 10
long_text
```

```python
print("Liczba tokenów w text:", len(embeddings._client.tokenizer.tokenize(text)))
print("Liczba tokenów w long text:", len(embeddings._client.tokenizer.tokenize(long_text)))
```

```python
v1 = embeddings.embed_query(long_text)
v2 = embeddings.embed_query(long_text + " Zupełnie inny temat: przepis na pierogi z kapustą.")

np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))  # ≈ 1.0 – dopisany tekst został ucięty
```

```python
v1 = embeddings.embed_query(text)
v2 = embeddings.embed_query(text + " ,")

np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))  # < 1.0 – znaczenie tekstu się zmieniło
```

Limity różnią się między modelami:

| model | maks. liczba tokenów |
|---|---|
| `all-mpnet-base-v2` | 384 |
| `paraphrase-multilingual-mpnet-base-v2` | 128 |
| `mmlw-roberta-base` | 512 |

## Podobieństwo między wektorami

Celem wektoryzacji tekstu w kontekście tworzenia systemu RAG jest umożliwienie wyszukiwania najbardziej adekwatnych fragmentów dokumentów względem promptu, który wpisał użytkownik. W tym celu niezbędne będzie korzystanie z metryk podobieństwa wektorów. 

Istnieją różne miary odległości wektorów, ale najczęściej stosowaną jest podobieństwo cosinusowe. Opiera się ona o cosinus kąta między wektorami w przestrzeni n-wymiarowej.



Porównywać podobieństwo wektorów powinniśmy tylko wtedy gdy zostały stworzone za pomocą tego samego modelu.

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

Poniższe funkcje zwracają dokumenty posortowane od najbardziej podobnego do zapytania. Działają z dowolnym obiektem `embeddings` z LangChaina (HuggingFace, OpenAI, ...).

Jeśli dokumenty zostały już wcześniej zwektoryzowane (np. duża liczba chunków), można przekazać gotowe wektory w `document_vectors`, żeby nie liczyć ich za każdym razem od nowa.

```python
def most_similar(query, documents, embeddings, top_k=None, document_vectors=None):
    """Zwraca listę par (podobieństwo, dokument) posortowaną malejąco."""
    query_vector = embeddings.embed_query(query)

    if document_vectors is None:
        document_vectors = embeddings.embed_documents(documents)

    similarities = [cosine_similarity(query_vector, vector) for vector in document_vectors]

    return sorted(zip(similarities, documents), reverse=True)[:top_k]


def print_most_similar(query, documents, embeddings, top_k=None, document_vectors=None):
    """Wyświetla ranking dokumentów najbardziej podobnych do zapytania."""
    results = most_similar(query, documents, embeddings, top_k, document_vectors)

    print(f"Zapytanie: {query}\n---")
    for i, (similarity, document) in enumerate(results, 1):
        print(f"{i}. [{similarity:.4f}] {document}\n")
```

```python
base = "Muszę zrobić zakupy w supermarkecie"
```

```python
sentences = [    
    "Dziś po pracy idę do sklepu na zakupy",
    "Wczoraj byłem na siłowni",
    "Zamówiłem pizzę na kolację",
    "Planuję kupić warzywa i owoce",
]
```

### `sentence-transformers/all-mpnet-base-v2`

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2", model_kwargs={"device": "cpu"})

print_most_similar(base, sentences, embeddings)
```

```python
most_similar(base, sentences, embeddings)
```

### `paraphrase-multilingual-mpnet-base-v2`

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-multilingual-mpnet-base-v2", model_kwargs={"device": "cpu"})

print_most_similar(base, sentences, embeddings)
```

### `sdadas/mmlw-roberta-base`

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sdadas/mmlw-roberta-base",
    model_kwargs={"device": "cpu"},
    query_encode_kwargs={"prompt": "zapytanie: "}
)

print_most_similar(base, sentences, embeddings)
```

### `text-embedding-ada-002`

```python
embeddings = OpenAIEmbeddings(model="text-embedding-ada-002")

print_most_similar(base, sentences, embeddings)
```
