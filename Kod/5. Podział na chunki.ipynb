# Podział na chunki

Celem podziału dokumentów na chunki jest zapis w bazie danych tylko fragmentów dokumentów. Te fragmenty będą później wyszukiwane (*retrieval*) i na ich podstawie będzie odbywać się generacja odpowiedzi.

## Dane

Każdy splitter z LangChaina udostępnia trzy metody do podziału:
- **`split_text()`** – przyjmuje tekst (`str`) i zwraca listę stringów (chunków bez metadanych),
- **`create_documents()`** – przyjmuje listę tekstów (`str`) oraz opcjonalnie listę metadanych i zwraca listę obiektów `Document`,
- **`split_documents()`** – przyjmuje listę obiektów `Document` (np. wynik loadera) i zwraca listę obiektów `Document`.

Dlatego przygotujemy dwa rodzaje danych: zwykły tekst z pliku `AI.txt` oraz listę dokumentów (stron) z wielostronicowego pliku `AI_long.pdf`.

````python
from langchain_community.document_loaders import TextLoader, PyMuPDFLoader
````

````python
text = TextLoader("documents/AI.txt").load()[0].page_content
print(text)
````

````python
documents = PyMuPDFLoader("documents/AI_long.pdf").load()
len(documents)  # jeden dokument na stronę
````

````python
[len(doc.page_content) for doc in documents]
````

````python
for doc in documents:
    print(doc.__repr__(), "\n\n===\n\n")
````

## Rodzaje splitterów

### `CharacterTextSplitter`

Dzieli tekst na podstawie określonego separatora (znaku lub ciągu znaków) a także oczekiwanej maksymalnej długości chunka.

````python
from langchain_text_splitters import CharacterTextSplitter
````

````python
splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=1000,  # maks. liczba znaków w chunku
    chunk_overlap=400  # liczba znaków wspólnych dla sąsiednich chunków
)
````

**`split_text()`**

````python
print(text)
````

````python
chunks = splitter.split_text(text)
len(chunks)
````

````python
for i, chunk in enumerate(chunks, 1):
    print(f"===== Chunk {i} jest typu {type(chunk)} ({len(chunk)} znaków) =====")
    print(chunk, "\n")
````

````python
chunk
````

**`create_documents()`**

Działa jak `split_text()`, ale zwraca obiekty `Document`. Przyjmuje listę tekstów oraz (opcjonalnie) listę słowników z metadanymi – po jednym dla każdego tekstu. Każdy chunk otrzymuje kopię metadanych tekstu, z którego powstał.

Każdy tekst z listy jest dzielony osobno – chunk nigdy nie łączy fragmentów dwóch tekstów, a overlap również działa tylko w obrębie jednego tekstu.

Przydaje się, gdy tekst nie pochodzi z loadera tylko jest stringiem a chcemy od razu otrzymać chunki klasy `Document` z metadanymi.

````python
chunks = splitter.create_documents(
    [text],
    metadatas=[{"source": "documents/AI.txt", "topic": "sztuczna inteligencja"}]
)
len(chunks)
````

````python
chunks[1].metadata
````

**`split_documents()`**

Każdy chunk jest obiektem `Document` i **dziedziczy metadane** dokumentu, z którego powstał (np. ścieżkę pliku i numer strony). Dzięki temu po wyszukaniu chunka wiemy, skąd pochodzi.

Każdy dokument jest dzielony osobno – chunk nigdy nie łączy tekstu z dwóch dokumentów (np. końca jednej strony i początku następnej), a overlap również działa tylko w obrębie jednego dokumentu.

Pod spodem `split_documents()` rozpakowuje dokumenty na listę tekstów i listę metadanych, a następnie wywołuje `create_documents()`, które dla każdego tekstu wywołuje `split_text()`:

```
split_documents(documents)  →  create_documents(texts, metadatas)  →  split_text(text)
```

````python
len(documents)
````

````python
chunks = splitter.split_documents(documents)
len(chunks)
````

````python
print(chunks[2].page_content)
````

````python
print(chunks[3].metadata)
````

````python
for i, chunk in enumerate(chunks):
    print(f"Chunk {i} | strona: {chunk.metadata['page']} | długość: {len(chunk.page_content)}")
````

Zwróć uwagę, że chunki mają po ponad 2000 znaków, mimo `chunk_size=1000`. Tekst wyciągnięty z PDF-a nie zawiera pustych linii (`"\n\n"`), a jedynie pojedyncze znaki `"\n"` na końcu każdej linii. Separator nie występuje, więc `CharacterTextSplitter` nie ma gdzie podzielić tekstu i każda strona trafia w całości do jednego chunka.

`chunk_size` nie jest więc twardym limitem – to ograniczenie rozwiązuje `RecursiveCharacterTextSplitter`, który w takiej sytuacji przechodzi do kolejnego separatora.

### `MarkdownHeaderTextSplitter`

Dzieli dokument w formacie Markdown na sekcje według nagłówków (`#`, `##`, …). Treść każdej sekcji trafia do osobnego chunka, a nagłówki, pod którymi się znajduje, są zapisywane w jego metadanych.

Plik `AI.md` zawiera ten sam tekst co `AI.txt`, ale z nagłówkami Markdown.

````python
from langchain_text_splitters import MarkdownHeaderTextSplitter
````

````python
md_text = TextLoader("documents/AI.md").load()[0].page_content
print(md_text)
````

````python
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#", "title"),  # (nagłówek, nazwa klucza w metadanych)
        ("##", "section"),
        ("###", "subsection"),
    ]
)
````

**`split_text()`**

W odróżnieniu od pozostałych splitterów `split_text()` zwraca tu od razu listę obiektów `Document` – metadane pochodzą z nagłówków. `MarkdownHeaderTextSplitter` nie ma metod `create_documents()` ani `split_documents()`.

````python
md_chunks = splitter.split_text(md_text)
len(md_chunks)
````

````python
for chunk in md_chunks:
    print(chunk.metadata, "| długość:", len(chunk.page_content))
````

````python
for chunk in md_chunks:
    print(chunk.page_content, "\n\n===\n\n")
````

**Łączenie z innym splitterem**

`MarkdownHeaderTextSplitter` nie kontroluje długości chunków – sekcja może być bardzo długa. Dlatego wynik często dzieli się dalej innym splitterem przy pomocy `split_documents()`. Chunki dziedziczą wtedy metadane z nagłówkami.

````python
splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", ". "],
    chunk_size=400,
    chunk_overlap=100
)

chunks = splitter.split_documents(md_chunks)
len(chunks)
````

````python
for chunk in chunks:
    print(chunk.metadata, "| długość:", len(chunk.page_content))
````

## Dobór `chunk_size` i `chunk_overlap`

Nie ma uniwersalnie najlepszych wartości – zależą od rodzaju dokumentów, pytań i modelu embeddingowego. Warto jednak pamiętać o kilku zasadach:

- **Za małe chunki** tracą kontekst – pojedyncze zdanie może nie zawierać informacji potrzebnej do odpowiedzi (np. kogo dotyczy).
- **Za duże chunki** zawierają wiele wątków naraz, przez co ich embedding staje się „rozmyty” i gorzej dopasowuje się do konkretnego pytania. Do modelu trafia też więcej niepotrzebnego tekstu (wyższy koszt, ryzyko zgubienia istotnej informacji).
- **Typowy punkt startowy** to ok. 500–1000 znaków (lub ok. 200–500 tokenów) na chunk.
- **Overlap** zapobiega przecięciu informacji na granicy chunków. Zwykle ustawia się go na ok. 10–20% `chunk_size` – zbyt duży overlap powoduje, że te same fragmenty powtarzają się w wielu chunkach.
- `chunk_size` musi mieścić się w limicie wejścia modelu embeddingowego (liczonym w tokenach).
- Najlepsze wartości dobiera się eksperymentalnie, na podstawie ewaluacji systemu RAG.

## Własne metadane chunków

Do metadanych chunków utworzonych przez `split_documents()` możemy dopisać własne informacje, np. długość chunka. Takie pola przydadzą się później do filtrowania wyników wyszukiwania.

````python
splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", ". "],
    chunk_size=1000,
    chunk_overlap=200
)

chunks = splitter.split_documents(documents)

for chunk in chunks:
    chunk.metadata["chunk_length"] = len(chunk.page_content)

chunks[0].metadata
````

## Własny splitter

Można stworzyć własny splitter tworząc klasę dziedziczącą po `TextSplitter`. Wystarczy po niej zaimplementować metodę `split_text()`. Metody `create_documents()` i `split_documents()` zostaną odziedziczone.

````python
from langchain_text_splitters import TextSplitter
````

````python
class MarkdownTableRowSplitter(TextSplitter):
    def __init__(self, rows_per_chunk=3):
        if type(rows_per_chunk) != int:
            raise ValueError("Invalid type of rows_per_chunk")
        
        if rows_per_chunk < 1:
            raise ValueError("rows_per_chunk must be greater than 1")

        super().__init__(chunk_size=rows_per_chunk, chunk_overlap=0)
        self.rows_per_chunk = rows_per_chunk

    def split_text(self, text):
        lines = [line.strip() for line in text.splitlines() if line.strip()]

        if len(lines) < 3 or not all(
            line.startswith("|") and line.endswith("|")
            for line in lines
        ):
            raise ValueError("Invalid table")

        header = lines[0]
        separator = lines[1]
        rows = lines[2:]
        chunks = []

        for start in range(0, len(rows), self.rows_per_chunk):
            chunk_rows = rows[start:start + self.rows_per_chunk]
            chunks.append("\n".join([header, separator, *chunk_rows]))

        return chunks
````

Przykładowa tabela jest częścią bazy wiedzy o planach abonamentowych. W systemie RAG użytkownik może zapytać np. „Który plan przechowuje dane przez rok?”.

````python
plans_table = """
| Plan | Cena miesięczna | Użytkownicy | Historia danych | Wsparcie |
|---|---|---|---|---|
| Starter | 49 zł | 3 | 30 dni | e-mail |
| Team | 99 zł | 10 | 90 dni | e-mail i czat |
| Pro | 199 zł | 30 | 1 rok | priorytetowe |
| Business | 399 zł | 100 | 3 lata | dedykowany opiekun |
| Enterprise | wycena | bez limitu | bez limitu | SLA 24/7 |
""".strip()

print(plans_table)
````

````python
splitter = MarkdownTableRowSplitter(rows_per_chunk=2)
chunks = splitter.split_text(plans_table)

for i, chunk in enumerate(chunks, 1):
    print(f"===== Chunk {i} =====")
    print(chunk, "\n")
````

Odziedziczona metoda `create_documents()` zamienia wynik na dokumenty LangChain i kopiuje do każdego chunka metadane źródła.

````python
documents = splitter.create_documents(
    [plans_table],
    metadatas=[{"source": "cennik.md", "content_type": "table"}],
)

documents
````
