## Prompt template, LLM Chain i structured output

2. Poniżej znajduje się kontekst (fragment "Pana Tadeusza") dotyczący moskiewskiego kapitana Rykowa. Wykorzystaj go, aby zadać modelowi pytanie o jego imię.

Utwórz `ChatPromptTemplate`, w którym zmiennymi będą kontekst i pytanie. W wiadomości systemowej poinstruuj model, aby odpowiadał zwięźle i wyłącznie na podstawie dostarczonego kontekstu. Połącz prompt z modelem w `chain` i wywołaj go. Porównaj odpowiedź modelu z oczekiwaną odpowiedzią.

---
(czas: 10 min.)

```python
context = """W izbie Sędziego ważne toczą się narady.
Bernardyn leżał w łóżku; zmordowany, blady
I skrwawiony, lecz całkiem zdrowy na umyśle,
Daje rozkazy, Sędzia wypełnia je ściśle.
Prosi Podkomorzego, przyzywa Klucznika,
Każe przywieść Rykowa, potem drzwi zamyka.
Godzinę całą trwały tajemne rozmowy,
Aż je przerwał kapitan Ryków tymi słowy,
Rzucając na stół kiesę ciężką dukatami:
«Państwo Lachy, już jest ta gadka między wami,
Że każdy Moskal złodziej: powiedźcież, kto spyta,
Że znaliście Moskala, który zwan Nikita
Nikitycz Ryków, rotny kapitan, miał osim
Medalów i trzy krzyże — to pamiętać prosim:
Ten medal za Oczaków, ten za Izmaiłów,
Ten za bitwę pod Nowi, ten za Prejsiż-Iłów ,
Tamten za Korsakowa sławną rejteradę
Z pod Zurich; a miał także i za męstwo szpadę,
Także od feldmarszałka trzy zadowolnienia,
Dwie pochwały cesarskie i cztery wspomnienia,
Wszystko na piśmie…»
"""
```

```python
question = "Jak miał na imię kapitan Ryków?"
```

```python
correct_answer = "Kapitan Ryków miał na imię Nikita."
```

```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
```

```python
messages = [
    SystemMessage("Odpowiedz zwięźle na pytanie dotyczące 'Pana Tadeusza'. Jeśli w kontekście nie będzie tej informacji, odpowiedz 'nie wiem'"),
    HumanMessagePromptTemplate.from_template("Kontekst:\n{context}\n\nPytanie: {question}")
]
```

```python
prompt = ChatPromptTemplate.from_messages(messages)
```

```python
llm = ChatOpenAI(model="openai/gpt-4o-mini").with_structured_output(ResponseFormat)

chain = prompt | llm
```

```python
response = chain.invoke({"context": context, "question": question})
```

3. Zmodyfikuj `chain` z poprzedniego zadania tak, aby korzystał ze *structured output*. Odpowiedź powinna zawierać następujące pola:
- pytanie
- odpowiedź
- wyjaśnienie (dokładny cytat z kontekstu, który uzasadnia odpowiedź)

---
(czas: 8 min.)

```python
from pydantic import BaseModel, Field

class ResponseFormat(BaseModel):
    question: str
    answer: str
    explanation: str = Field(description="Fragment kontekstu, dokładny cytat w którym występuje odpowiedz")
```
