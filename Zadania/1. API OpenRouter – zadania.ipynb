```python
questions = [
    "Pomiędzy kim trwa spór o zamek Horeszków?",
    "Jak miał na imię kapitan Ryków?",
    "Wymień wszystkie zioła które znajdowały się w donicach na oknach w domu Tadeusza.",
    "'Kapitanie! nie splamisz czci twojej pardonem. Dałeś próby, rycerzu nieszczęsny, lecz mężny' – kto wypowiada te słowa?",
    "Od kogo Sędzia otrzymał w darze kryształowy kufel?",
    "Czym się różnili Dobrzyńscy od reszty Litwinów?",
    "Co mówi Sędzia o wódce gdańskiej?",
    "Jak Domejko i Dowejko nazwali karczmę którą zbudowali po zakończeniu sporu?",
    "Kto wypowiedział słowa 'Kometa czasem wojny, czasem wróży kłótnie!'",
    "Jakie ustawy znajdują się w żółtej księdze?"
]
```

```python
correct_answers = [
    "Spór o zamek trwa pomiędzy rodami Sopliców i Horeszków.",
    "Kapitan Ryków miał na imię Nikita.",
    "Te zioła to: geranium, lewkonija, astry i fiołki.",
    "Te słowa wypowiada Podkomorzy.",
    "Sędzia otrzymał go od księdza Robaka.",
    "Dobrzyńscy wyróżniali się językiem oraz wzrostem.",
    "Mówi, że wódka gdańska to napój miły dla Polaka.",
    "Nazwali tę karczmę 'Niedźwiadek'.",
    "Te słowa wypowiedział Wojski.",
    "W żółtej księdze znajdują się ustawy wojenne."
]
```

```python
from openai import OpenAI
import os
from dotenv import load_dotenv
from pydantic import BaseModel

load_dotenv()
```

```python
client = OpenAI()
```

```python
actual_answers = []
```

```python
for question in questions:
    messages = [
        {"role": "system", "content": "Odpowiadaj na pytanie dotyczące 'Pana Tadeusza'"},
        {"role": "user", "content": question}
    ]
    response = client.responses.create(input=messages, model="openai/gpt-4o-mini")
    actual_answers.append(response.output_text)
```

---
---
---

```python
class EvaluateAnswer(BaseModel):
    question: str
    actual_answer: str
    expected_answer: str
    evaluation: bool
```

```python
results = []
```

```python
for question, actual_answer, expected_answer in zip(questions, actual_answers, correct_answers):
    messages = [
        {"role": "system", "content": "Dokonaj ewaluacji rzeczywistej odpowiedzi modelu porównując ją do oczekiwanej"},
        {"role": "user", "content": f"Pytanie: {question}, rzeczywista odpowiedź: {actual_answer}, oczekiwana odpowiedz: {expected_answer}"}
    ]
    response = client.responses.parse(input=messages, model="openai/gpt-4o-mini", text_format=EvaluateAnswer)
    results.append(response.output_parsed.model_dump())
```

```python
results
```

```python
[item["evaluation"] for item in results]
```
